Problem 1 - Reverse the String

```python

def reverse_string(text:str) -> str:
  return text[::-1]

# when single use

reverse_string1 = lambda text : text [::-1]

print(reverse_string("Hello World"))
print(reverse_string1("Hello World"))
```

problem 2 - palindrome

```python
def palindrome(text:str) -> bool:
  palin = "".join(ch.casefold() for ch in text if ch.isalnum())
  return palin == palin[::-1]


print(palindrome("Madam"))
print(palindrome("Hello World"))

# Array

def array_palindrome(arr) -> bool:
  return arr == arr[::-1]

print(array_palindrome([1,2,3,2,1]))
print(array_palindrome([1,2,3,4,5]))
```

Problem 3 - charecter frequency

```python
from collections import Counter

def freq(arr):
  return dict(Counter(arr))

print(freq([1,2,3,2,1]))
print(freq([1,2,3,4,5]))

# Inside the counter:

def count(arr):
  freq = {}
  for x in arr:
    freq[x] = freq.get(x, 0) + 1
  print(freq)

count([1,2,3,2,1])
```

Problem 4 - 1st non repeat val

```python
def first_non_repeat(arr):
  order = {}
  freq = {}

  for i in arr:
    order[i] = arr
    freq[i] = freq.get(i, 0) + 1

  for i in order:
    if freq[i] == 1:
      return i

print(first_non_repeat([1,2,3,5,6,2,4,5,6]))

```

Problem 5 - Detect Duplicates.

```python
from collections import Counter

def duplicate(arr):
  dup = [item for item,count in Counter(arr).items() if count > 1]
  return dup

print(duplicate([1,3,3,2,1,4,5,4,5,6]))
```

Problem 6 - Two sum

```python
def twosum(arr, target):
    seen = {}

    for index, value in enumerate(arr):
        bal = target - value

        if bal in seen:
            return [seen[bal], index]

        seen[value] = index

    return []
```

Problem 8 - Largest and Second largest distinct.

```python
def largest_second_largest(arr):
  first=second=None

  for value in arr:

    if first is None or second is None:
      continue

    if first is None or value > first:
      first, second = value, first

    elif second is None or value > second:
      second = value

  return first, second
```

Problem 9 - Move Zeros to the end.

```python
def move_zero_to_end(arr):
  new_arr = [new_arr.append(x) for x in arr if x!= 0]
  new_arr.extend([0] * (len(arr) - len(new_arr)))
  return new_arr
```

Problem 11 - Merge two sorted lists

```python
def merge(a, b):
  i=j=0
  merged = []

  while i < len(a) and j < len(b):
    if a[i] <= b[j]:
      merged.append(a[i])
      i += 1
    else:
      merged.append(b[j])
      j += 1

  merged.extend(a[i:])
  merged.extend(b[j:])
```

Problem 12 - Binary Search

```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = left + (right - left) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

problem 13 - Factorial

Problem 14 - Fibonacci Number

```python
def fibonacci(num):
  a, b = 0, 1
  for _ in range(0, num):
     a,b = b, a+b
     return a
```

problem 17 - Most Frequent

```python
from collections import Counter

def most_freq(arr):
    freq = Counter(arr)
    most_frequent = arr[0]

    for i in freq:
        if freq[i] > freq[most_frequent]:
            most_frequent = i

    return most_frequent
```
